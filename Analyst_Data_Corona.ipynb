```python
import pandas as pd
corona = pd.read_csv('/content/drive/MyDrive/BelajarAnalyst/Corona/CovidLive.csv')
corona.head(1)
```

```python
sortCorona=corona.rename(columns={
    'Country,\nOther':'Country',
    'Total\nCases':'TotalCases',
    'Total\nDeaths':'TotalDeaths',
    'New\nDeaths':'NewDeaths',
    'Total\nRecovered':'TotalRecovered',
    'Active\nCases':'ActiveCases',
    'Serious,\nCritical':'Serious,\nCritical',
    'Tot Cases/\n1M pop':'TotalCases_1Pop',
    'Deaths/\n1M pop':'Deaths_1MPop',
    'Total\nTests':'Total Test',
    'Tests/\n1M pop':'Test_1MPop',
    'Population':'Population'})
sortCorona = sortCorona[['Country','TotalCases','TotalDeaths','TotalRecovered']]
sortCorona.head(3)
```

```python
sortCorona=sortCorona.rename(columns={'Country,\nOther':'Country', 'Total\nCases':'TotalCases'})
sortCorona['TotalCases'] = sortCorona['TotalCases'].str.replace(',', '').astype(int)
```

```python
sortterkecil=sortCorona.sort_values(by = 'TotalCases',ascending=True,inplace=True)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Pastikan kolom 'TotalCases' dalam format numerik
sortCorona['TotalCases'] = sortCorona['TotalCases'].replace(',', '', regex=True).astype(int)

# Ambil 10 negara dengan jumlah kasus tertinggi
top_countries = sortCorona.sort_values(by='TotalCases', ascending=False).head(10)

# Data untuk pie chart
countries = top_countries['Country']
cases = top_countries['TotalCases']

# Buat pie chart
plt.figure(figsize=(8, 8))  # Ukuran figure agar proporsional
plt.pie(cases, labels=[f"{c} ({v:,})" for c, v in zip(countries, cases)], startangle=140, labeldistance=1.1)


# Tambahkan judul
plt.title('Persentase Total Kasus COVID-19 per Negara')

# Tampilkan plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Pastikan 'TotalDeaths' dalam format numerik (hapus koma jika ada)
sortCorona['TotalDeaths'] = sortCorona['TotalDeaths'].replace(',', '', regex=True).astype(float)

# Ambil 10 negara dengan jumlah kematian tertinggi
top_countries = sortCorona.sort_values(by='TotalDeaths', ascending=False).head(10)

# Data untuk pie chart
countries = top_countries['Country']
cases = top_countries['TotalDeaths']

# Buat pie chart
plt.figure(figsize=(8, 8))  # Ukuran figure agar proporsional
plt.pie(cases, labels=[f"{c} ({int(v):,})" for c, v in zip(countries, cases)], startangle=140, labeldistance=1.1)

# Tambahkan judul
plt.title('Persentase Total Kematian Kasus COVID-19 per Negara')

# Tampilkan plot
plt.show()
```
